# Python Getting Started

```python
print("Hello world")
```

```python
# Python Version
import sys
print(sys.version)
```

```python
# Text in Python must be inside quotes. We can use either " double quotes or ' single quotes:
print("This will work!")
print('This will also work!')
```

```python
# By default, the print() function ends with a new line. If we want to print multiple words on the same line, we can use the end parameter:

print("Hello World!", end=" ")
print("I will print on the same line.")
```

```python
# We can also use the print() function to display numbers:
print(2)
print(10)
```

```python
# We can also do math inside the print() function:
print(10 + 30)
print(40 - 20)
print(7 * 9)
print(10 / 5)
```

```python
# You can combine text and numbers in one output by separating them with a comma:
print("I am", 35, "years old.")
```
