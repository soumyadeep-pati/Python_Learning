# Getting the Data Type

```python
# We can get the data type of any object by using the type() function:
x = 5
y = "Hello World"
print(type(x))
print(type(y))
```

```python
# If We want to specify the data type, you can use the following constructor functions:
a = str("Hello World")
b = int(20)
c = float(20.5)
d = complex(1j)
e = list(("apple", "banana", "cherry"))
print(type(a))
print(type(b))
print(type(c))
print(type(d))
print(type(e))
```

# Python Numbers

```python
# There are three numeric types in Python: int, float, complex

x = 1    # int
y = 2.8  # float
z = 1j   # complex

print(type(x))
print(type(y))
print(type(z))
```

# Python Casting

```python
x = int(1)   # x will be 1
y = int(2.8) # y will be 2
z = int("3") # z will be 3
print(x)
print(y)
print(z)
```

```python
x = float(1)     # x will be 1.0
y = float(2.8)   # y will be 2.8
z = float("3")   # z will be 3.0
w = float("4.2") # w will be 4.2
print(x)
print(y)
print(z)
```

```python
x = str("s1") # x will be 's1'
y = str(2)    # y will be '2'
z = str(3.0)  # z will be '3.0'
print(x)
print(y)
print(z)
```
