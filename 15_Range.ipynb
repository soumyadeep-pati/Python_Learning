# Python range

range(start, stop, step)

```python
# Create a range of numbers from 0 to 9:
x = range(10)
```

```python
# Create a range of numbers from 3 to 9:
x = range(3, 10)
```

```python
# Create a range of numbers from 3 to 9:
x = range(3, 10, 2)
```

```python
# Iterate over each value in a range:
for i in range(10):
    print(i)
```

```python
# Convert different ranges to lists:

print(list(range(5)))
print(list(range(1, 6)))
print(list(range(5, 20, 3)))
```

```python
# Slicing Ranges
r = range(10)
print(r[2])
print(r[:3])
```

```python
# Membership Testing
r = range(0, 10, 2)
print(list(range(0, 10, 2)))
print(6 in r)
print(7 in r)
```

```python
# Length
r = range(0, 10, 2)
for i in list(r):
    print(i)
print()
print(len(r))
```
