#  Duplicates and Data Cleaning

Real-world datasets are often messy.

Problems can include:

- Duplicate rows
- Duplicate records
- Extra spaces
- Inconsistent capitalization
- Incorrect values
- Wrong data types
- Invalid values
- Inconsistent category names

In this notebook, we will learn how to identify and
fix these common problems using Pandas.

```python
# Import Pandas
import pandas as pd
```

```python
# Create a dataset containing some messy data

data = {
    "Name": [
        "Rahul",
        "Priya",
        "Amit",
        "Sneha",
        "Rahul",
        "  Rohan  "
    ],
    
    "Age": [
        20,
        21,
        22,
        20,
        20,
        23
    ],
    
    "Department": [
        "CSE",
        "ECE",
        "cse",
        "CSE ",
        "CSE",
        "ece"
    ],
    
    "Marks": [
        85,
        90,
        76,
        92,
        85,
        68
    ]
}

df = pd.DataFrame(data)

print(df)
```

## Check Duplicate Rows

```python
# Check which rows are duplicates
print(df.duplicated())
```

## Count Duplicate Rows

```python
# Count the number of duplicate rows
print(df.duplicated().sum())
```

## Display Duplicate Rows

```python
# Display only duplicate rows
duplicates = df[df.duplicated()]

print(duplicates)
```

## Remove Duplicate Rows


```python
# Remove duplicate rows
df_clean = df.drop_duplicates()

print(df_clean)
```

## Reset the Index

```python
# Reset the index after removing duplicates
df_clean = df_clean.reset_index(drop=True)

print(df_clean)
```

## Duplicate Based on Specific Columns

```python
# Check duplicates based on Name and Age
print(df.duplicated(subset=["Name", "Age"]))
```

## Remove Duplicates Based on Specific Columns

```python
# Remove duplicates based on Name and Age
df_unique = df.drop_duplicates(
    subset=["Name", "Age"]
)

print(df_unique)
```

## Remove Spaces from Text

```python
# Remove leading and trailing spaces from Name
df["Name"] = df["Name"].str.strip()

print(df)
```

## Convert Text to Lowercase

```python
# Convert Department values to lowercase
df["Department"] = df["Department"].str.lower()

print(df)
```

## Remove Spaces and Convert Case

```python
# Remove spaces and convert Department to uppercase
df["Department"] = (
    df["Department"]
    .str.strip()
    .str.upper()
)

print(df["Department"])
```

## Check Categories After Cleaning

```python
# Check unique Department values
print(df["Department"].unique())
```

```python
print(df["Department"].value_counts())
```

## Replace Incorrect Values

```python
# Replace incorrect category
df["Department"] = df["Department"].replace(
    "CS",
    "CSE"
)
```

```python
# multiple replacements:

df["Department"] = df["Department"].replace({
    "CS": "CSE",
    "Computer Science": "CSE"
})
```

## Detect Invalid Numerical Values

```python
# Create an example with an invalid age
df.loc[0, "Age"] = -5

print(df)
```

```python
# Find rows with invalid ages
invalid_age = df[df["Age"] < 0]

print(invalid_age)
```

## Replace Invalid Values with Missing Values

```python
# Replace invalid ages with NaN
df.loc[df["Age"] < 0, "Age"] = None

print(df)
```

```python
df["Age"] = df["Age"].fillna(
    df["Age"].median()
)
df
```

## Detect Invalid Marks

```python
# Find invalid marks
invalid_marks = df[
    (df["Marks"] < 0) |
    (df["Marks"] > 100)
]

print(invalid_marks)
```

## Convert Text Numbers to Numeric.

```python
# Create an example DataFrame
salary_df = pd.DataFrame({
    "Salary": ["25000", "30000", "35000", "unknown"]
})

print(salary_df)
print(salary_df.dtypes)
```

```python
# Convert Salary to numeric
salary_df["Salary"] = pd.to_numeric(
    salary_df["Salary"],
    errors="coerce"
)

print(salary_df)
```

## Remove Duplicate Names

```python
# Remove duplicate names
df = df.drop_duplicates(
    subset=["Name"]
)

print(df)
```

## Rename Columns for Consistency

```python
# Example DataFrame
messy_columns = pd.DataFrame({
    "Student Name": ["A", "B"],
    "student_age": [20, 21],
    "MARKS": [80, 90]
})

print(messy_columns)
```

```python
# Standardize column names
messy_columns = messy_columns.rename(columns={
    "Student Name": "student_name",
    "student_age": "age",
    "MARKS": "marks"
})

print(messy_columns)
```

## Remove Extra Spaces from Column Names

```python
# Remove extra spaces from all column names
df.columns = df.columns.str.strip()

print(df.columns)
```

## Convert Column Names to Lowercase

```python
df.columns = df.columns.str.strip().str.lower()
df
```

## A Practical Cleaning Pipeline

```python
# Create a messy dataset

data = {
    "Name": [" Rahul ", "Priya", "Amit", "Rahul"],
    "Age": [20, 21, -5, 20],
    "Department": ["CSE", "cse", " CSE ", "CSE"],
    "Marks": [85, 90, 76, 85]
}

df = pd.DataFrame(data)

print(df)
```

```python
# Step 2: Identify invalid ages

print(df[df["Age"] < 0])
```

```python
# Replace invalid ages with NaN
df.loc[df["Age"] < 0, "Age"] = None
df
```

```python
# Step 4: Fill missing Age values using the median

df["Age"] = df["Age"].fillna(
    df["Age"].median()
)

print(df)
```

```python
# Step 5: Remove duplicate records

df = df.drop_duplicates(
    subset=["Name", "Age", "Department", "Marks"]
)

# Reset the index
df = df.reset_index(drop=True)

print(df)
```

```python
# Step 6: Final inspection

print("Shape:")
print(df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nUnique departments:")
print(df["Department"].unique())

print("\nData types:")
print(df.dtypes)
```
