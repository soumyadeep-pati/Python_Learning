```python
# Create a sample customer dataset

data = {
    "Age": [22, 35, 28, 45, 30, 40, 25, 50],
    "Income": [25000, 60000, 45000, 80000, 50000, 75000, 30000, 90000],
    "City": [
        "Delhi",
        "Mumbai",
        "Delhi",
        "Kolkata",
        "Mumbai",
        "Delhi",
        "Kolkata",
        "Mumbai"
    ],
    "Experience": [1, 10, 5, 15, 7, 12, 3, 20],
    "Purchased": [
        "No",
        "Yes",
        "Yes",
        "Yes",
        "No",
        "Yes",
        "No",
        "Yes"
    ]
}

df = pd.DataFrame(data)

print(df)
```

## Let's Add Some Missing Values

```python
# Create a copy so the original dataset remains unchanged
df_ml = df.copy()

# Introduce some missing values
df_ml.loc[2, "Income"] = None
df_ml.loc[4, "City"] = None

print(df_ml)
```

```python
# Check missing values
print(df_ml.isnull().sum())
```

## Handle Numerical Missing Data

```python
# Fill missing Income values with the median
df_ml["Income"] = df_ml["Income"].fillna(
    df_ml["Income"].median()
)

print(df_ml)
```

```python
# Verify missing values
print(df_ml.isnull().sum())
```

## Handle Categorical Missing Data

```python
# Fill missing City with the most common city
df_ml["City"] = df_ml["City"].fillna(
    df_ml["City"].mode()[0]
)

print(df_ml)
```

```python
# Verify that missing values have been handled
print(df_ml.isnull().sum())
```

## Check Duplicate Rows

```python
# Check for duplicate rows
print(df_ml.duplicated().sum())
```

```python
df_ml = df_ml.drop_duplicates()
```

## Understand Features and Target

## Create X

```python
# Select the input features
X = df_ml[
    ["Age", "Income", "City", "Experience"]
]

print(X)
```

```python
# Check the type of X
print(type(X))
```

## Create y

```python
# Select the target variable
y = df_ml["Purchased"]

print(y)
```

#  Understanding X and y

Suppose our dataset is:

| Age | Income | City | Experience | Purchased |
|-----|--------|------|------------|-----------|
| 22 | 25000 | Delhi | 1 | No |
| 35 | 60000 | Mumbai | 10 | Yes |
| 28 | 45000 | Delhi | 5 | Yes |

We want the model to learn:

    Age + Income + City + Experience
                     ↓
                Purchased

Therefore:

X = input features

y = target/output

In code:

X = df[["Age", "Income", "City", "Experience"]]

y = df["Purchased"]

## Check X and y Shapes

```python
# Check the shape of X
print("X shape:", X.shape)

# Check the shape of y
print("y shape:", y.shape)
```

## Numerical vs Categorical Columns

```python
# Select numerical columns
numerical_columns = X.select_dtypes(
    include=["int64", "float64"]
)

print(numerical_columns)
```

## Select Categorical Columns

```python
# Select categorical columns
categorical_columns = X.select_dtypes(
    include=["object"]
)

print(categorical_columns)
```

## Simple Encoding with map()

```python
# Convert the binary target into numerical values

y_encoded = y.map({
    "No": 0,
    "Yes": 1
})

print(y_encoded)
```

## Check the Target Distribution

```python
# Count the target classes
print(y.value_counts())
```

```python
# Count encoded target classes
print(y_encoded.value_counts())
```

## Encoding a Categorical Feature

### Simple One-Hot Encoding with Pandas

```python
# Convert City into one-hot encoded columns
city_encoded = pd.get_dummies(
    X["City"],
    dtype=int
)

print(city_encoded)
```

## Combine Encoded Data with Numerical Features

```python
# Select numerical features
X_numeric = X[
    ["Age", "Income", "Experience"]
]

# Encode City
X_city = pd.get_dummies(
    X["City"],
    dtype=int
)

# Combine numerical and encoded features
X_final = pd.concat(
    [X_numeric, X_city],
    axis=1
)

print(X_final)
```

## Check the Final Dataset

```python
# Check the final feature data types
print(X_final.dtypes)
```

## Check Final X and y

```python
# Display the final features
print("X:")
print(X_final)

print("\ny:")
print(y_encoded)
```
