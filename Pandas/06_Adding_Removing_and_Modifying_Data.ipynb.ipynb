#  Adding, Removing and Modifying Data

In this notebook, we will learn how to:

1. Add new columns
2. Modify existing columns
3. Create columns from calculations
4. Rename columns
5. Remove columns
6. Remove rows
7. Change data types
8. Replace values
9. Apply functions to columns

These operations are commonly used during data cleaning
and Machine Learning preprocessing.

```python
# Import Pandas
import pandas as pd
```

```python
# Create a sample student dataset

data = {
    "Name": ["Rahul", "Priya", "Amit", "Sneha", "Rohan"],
    "Age": [20, 21, 22, 20, 23],
    "Marks": [85, 90, 76, 92, 68],
    "Department": ["CSE", "ECE", "CSE", "CSE", "ECE"]
}

df = pd.DataFrame(data)

print(df)
```

```python
# Add a new column called Passed
df["Passed"] = True

print(df)
```

## Create a Column Using a Condition

```python
# Create Passed based on Marks
df["Passed"] = df["Marks"] >= 40

print(df)
```

## Create a Column from Arithmetic

```python
# Add 5 bonus marks to every student
df["Final_Marks"] = df["Marks"] + 5

print(df)
```

## Create a Column Using Multiple Columns

```python
# Calculate the bonus received by each student
df["Bonus"] = df["Final_Marks"] - df["Marks"]

print(df)
```

## Create a Column Using a Formula

```python
# Calculate percentage
df["Percentage"] = (df["Marks"] / 100) * 100

print(df)
```

## Modify an Existing Column

```python
# Increase every student's marks by 2
df["Marks"] = df["Marks"] + 2

print(df)
```

## Rename a Column

```python
# Rename the Marks column
df = df.rename(columns={"Marks": "Score"})

print(df)
```

## Rename Multiple Columns

```python
# Rename multiple columns
df = df.rename(columns={
    "Name": "Student_Name",
    "Age": "Student_Age"
})

print(df)
```

## Remove a Column

```python
# Remove the Bonus column
df = df.drop("Bonus", axis=1)

print(df)
```

## Remove Multiple Columns

```python
# Remove multiple columns
df = df.drop(
    ["Final_Marks", "Percentage"],
    axis=1
)

print(df)
```

## Remove a Row

```python
# Remove the row with index 2
df = df.drop(2, axis=0)

print(df)
```

## Reset the Index

```python
# Reset the index after removing a row
df = df.reset_index(drop=True)

print(df)
```

## Replace Values

```python
# Replace CSE with Computer Science
df["Department"] = df["Department"].replace(
    "CSE",
    "Computer Science"
)

print(df)
```

## Replace Multiple Values

```python
# Replace multiple values
df["Department"] = df["Department"].replace({
    "ECE": "Electronics",
    "Computer Science": "CSE"
})

print(df)
```

## Changing Data Types with astype()

```python
# Create a DataFrame containing numbers stored as strings

data = {
    "Age": ["20", "21", "22", "23"],
    "Marks": ["85", "90", "76", "92"]
}

numbers_df = pd.DataFrame(data)

print(numbers_df)
print(numbers_df.dtypes)
```

## Convert String to Integer

```python
# Convert Age and Marks to integers

numbers_df["Age"] = numbers_df["Age"].astype(int)
numbers_df["Marks"] = numbers_df["Marks"].astype(int)

print(numbers_df.dtypes)
```

## Convert to Float

```python
# Convert Marks to floating-point numbers
numbers_df["Marks"] = numbers_df["Marks"].astype(float)

print(numbers_df)
print(numbers_df.dtypes)
```

## apply() — Applying a Function

```python
# Function to classify marks

def grade(marks):
    if marks >= 90:
        return "A"
    elif marks >= 80:
        return "B"
    elif marks >= 70:
        return "C"
    else:
        return "D"
```

```python
# Apply the grade function to every value in Marks
df["Grade"] = df["Score"].apply(grade)

print(df)
```

## Lambda with apply()

```python
# Create a new column using a lambda function
df["Double_Score"] = df["Score"].apply(
    lambda x: x * 2
)

print(df)
```

## Conditional Modification

```python
# Replace scores below 70 with 0
df.loc[df["Score"] < 70, "Score"] = 0

print(df)
```

```python
# Add 5 bonus marks to students who scored above 80
df.loc[df["Score"] > 80, "Score"] += 5

print(df)
```
