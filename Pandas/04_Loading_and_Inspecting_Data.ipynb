#  Loading and Inspecting Data with Pandas

In Machine Learning, we usually start with a dataset.

Datasets can come from:

- CSV files
- Excel files
- Databases
- APIs
- Other data sources

In this notebook, we will focus on CSV files.

We will learn:

1. How to load a CSV file
2. How to inspect the dataset
3. How to understand rows and columns
4. How to check data types
5. How to find missing values
6. How to find duplicate rows
7. How to generate basic statistics

```python
# Import Pandas
import pandas as pd
```

```python
# Create sample student data

data = {
    "Name": ["Rahul", "Priya", "Amit", "Sneha", "Rohan"],
    "Age": [20, 21, 22, 20, 23],
    "Marks": [85, 90, 76, 92, 68]
}

df = pd.DataFrame(data)

print(df)
```

```python
# Save the DataFrame as a CSV file
df.to_csv("students.csv", index=False)
```

```python
# Load the CSV file into a DataFrame
df = pd.read_csv("students.csv")

print(df)
```

```python
# Display the first five rows
print(df.head())
```

```python
# Display the first three rows
print(df.head(3))
```

```python
# Display the last five rows
print(df.tail())
```

```python
# Get the number of rows and columns
print(df.shape)
```

```python
# Display column names
print(df.columns)
```

```python
# Display data types of all columns
print(df.dtypes)
```

```python
# Display a summary of the DataFrame
df.info()
```

```python
# Generate statistical information
print(df.describe())
```

```python
# Check for missing values in every column
print(df.isnull().sum())
```

```python
# Create a copy so we don't modify our original DataFrame
df_missing = df.copy()

# Create a missing value
df_missing.loc[2, "Marks"] = None

print(df_missing)
```

```python
# Count missing values in each column
print(df_missing.isnull().sum())
```

```python
df.isna().sum()
```

```python
df_missing.isna().sum()
```

```python
# Check for duplicate rows
print(df.duplicated())
```

```python
# Count the number of duplicate rows
print(df.duplicated().sum())
```

```python
# Create a copy
df_duplicate = df.copy()

# Add a duplicate of the first row
df_duplicate.loc[len(df_duplicate)] = df_duplicate.iloc[0]

print(df_duplicate)
```

```python
print(df_duplicate.duplicated().sum())
```

```python
# Create a categorical column
df["Department"] = [
    "CSE",
    "ECE",
    "CSE",
    "CSE",
    "ECE"
]

print(df)
```

```python
# Display unique departments
print(df["Department"].unique())
```

```python
# Count the number of unique departments
print(df["Department"].nunique())
```

```python
# Count how many students belong to each department
print(df["Department"].value_counts())
```

```python
# Check the average marks
print(df["Marks"].mean())
```

```python
# Find the highest marks
print(df["Marks"].max())
```

```python
# Find the highest marks
print(df["Marks"].max())
```
