## Why do we write `pd`?

`pd` is the standard alias used for Pandas.

Instead of writing:

pandas.DataFrame()

we can write:

pd.DataFrame()

This makes our code shorter and easier to read.

```python
# Check the installed Pandas version
print(pd.__version__)
```

# 1. Pandas Series

A Series is a one-dimensional labeled data structure.

You can think of it as:

- A single column
- A list with labels/indexes

Example:

    10
    20
    30
    40

Pandas automatically gives each value an index.

```python
# Create a Pandas Series
data = pd.Series([10, 20, 30, 40])

print(data)
```

```python
# Access the first value
print(data[0])
```

```python
# Access the third value
print(data[2])
```

```python
# Create a Series with custom indexes
marks = pd.Series(
    [85, 90, 78],
    index=["Math", "Python", "DBMS"]
)

print(marks)
```

```python
# Access the Python marks
print(marks["Python"])
```

# 4. Pandas DataFrame

A DataFrame is a two-dimensional table.

Think of it like an Excel spreadsheet or a database table.

It contains:

- Rows
- Columns
- Indexes
- Values

Example:

| Name | Age | Marks |
|------|-----|-------|
| Rahul | 20 | 85 |
| Priya | 21 | 90 |
| Amit | 20 | 78 |

This type of structure is extremely common in Machine Learning datasets.

```python
# Create a dictionary containing student information
student_data = {
    "Name": ["Rahul", "Priya", "Amit"],
    "Age": [20, 21, 20],
    "Marks": [85, 90, 78]
}

# Convert the dictionary into a DataFrame
df = pd.DataFrame(student_data)

# Display the DataFrame
print(df)
```

```python
# Display the first 5 rows
print(df.head())
```

```python
# Display the last 5 rows
print(df.tail())
```

```python
# Get the number of rows and columns
print(df.shape)
```

```python
# Get the column names
print(df.columns)
```

```python
# Select the Age column
print(df["Age"])
```

```python
# You can also write:
print(df.Age)
```

```python
# Select multiple columns
print(df[["Name", "Marks"]]) # Notice the double brackets:
```
