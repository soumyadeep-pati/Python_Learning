#  Merging and Combining Data

In real-world projects, information is often stored in
multiple DataFrames or files.

For example:

Student information:

    Student_ID
    Name
    Age

Marks information:

    Student_ID
    Marks
    Grade

We need to combine these datasets.

In this notebook, we will learn:

1. `concat()`
2. `merge()`
3. Inner join
4. Left join
5. Right join
6. Outer join
7. Merging using multiple columns

```python
# Import Pandas
import pandas as pd
```

```python
# Student information

students = pd.DataFrame({
    "Student_ID": [101, 102, 103, 104],
    "Name": ["Rahul", "Priya", "Amit", "Sneha"],
    "Age": [20, 21, 22, 20]
})

print(students)
```

```python
# Student marks information

marks = pd.DataFrame({
    "Student_ID": [101, 102, 103, 105],
    "Marks": [85, 90, 76, 88],
    "Grade": ["B", "A", "C", "B"]
})

print(marks)
```

## concat()

```python
# Create two small DataFrames

df1 = pd.DataFrame({
    "Name": ["Rahul", "Priya"],
    "Marks": [85, 90]
})

df2 = pd.DataFrame({
    "Name": ["Amit", "Sneha"],
    "Marks": [76, 92]
})

print(df1)
print()
print(df2)
```

```python
# Combine the DataFrames row-wise

result = pd.concat([df1, df2])

print(result)
```

## Reset Index After concat()

```python
# Combine DataFrames and create a new index

result = pd.concat(
    [df1, df2],
    ignore_index=True
)

print(result)
```

## Concatenating Columns

```python
# Create two DataFrames with the same number of rows

df1 = pd.DataFrame({
    "Name": ["Rahul", "Priya", "Amit"]
})

df2 = pd.DataFrame({
    "Marks": [85, 90, 76]
})

# Combine column-wise
result = pd.concat([df1, df2], axis=1)

print(result)
```

## Basic merge()

```python
# Recreate the original DataFrames

students = pd.DataFrame({
    "Student_ID": [101, 102, 103, 104],
    "Name": ["Rahul", "Priya", "Amit", "Sneha"],
    "Age": [20, 21, 22, 20]
})

marks = pd.DataFrame({
    "Student_ID": [101, 102, 103, 105],
    "Marks": [85, 90, 76, 88],
    "Grade": ["B", "A", "C", "B"]
})
```

```python
# Merge using Student_ID

result = pd.merge(
    students,
    marks,
    on="Student_ID"
)

print(result)
```

## Inner Join

```python
# Perform an inner join

result = pd.merge(
    students,
    marks,
    on="Student_ID",
    how="inner"
)

print(result)
```

## Left Join

```python
# Perform a left join

result = pd.merge(
    students,
    marks,
    on="Student_ID",
    how="left"
)

print(result)
```

## Right Join

```python
# Perform a right join

result = pd.merge(
    students,
    marks,
    on="Student_ID",
    how="right"
)

print(result)
```

## Outer Join

```python
# Perform an outer join

result = pd.merge(
    students,
    marks,
    on="Student_ID",
    how="outer"
)

print(result)
```

## Merge Using Different Column Names

```python
students = pd.DataFrame({
    "Student_ID": [101, 102, 103],
    "Name": ["Rahul", "Priya", "Amit"]
})

marks = pd.DataFrame({
    "ID": [101, 102, 103],
    "Marks": [85, 90, 76]
})
```

```python
# Merge using columns with different names

result = pd.merge(
    students,
    marks,
    left_on="Student_ID",
    right_on="ID"
)

print(result)
```

## Remove Duplicate Key Column

```python
# Remove the duplicate ID column

result = result.drop("ID", axis=1)

print(result)
```

## Merge Using Multiple Columns

```python
students = pd.DataFrame({
    "Student_ID": [101, 101, 102, 102],
    "Semester": [1, 2, 1, 2],
    "Name": ["Rahul", "Rahul", "Priya", "Priya"]
})

marks = pd.DataFrame({
    "Student_ID": [101, 101, 102, 102],
    "Semester": [1, 2, 1, 2],
    "Marks": [80, 85, 90, 95]
})

result = pd.merge(
    students,
    marks,
    on=["Student_ID", "Semester"]
)

print(result)
```
