#  Selecting and Filtering Data with Pandas

In this notebook, we will learn how to:

1. Select columns
2. Select rows
3. Use `loc`
4. Use `iloc`
5. Filter using conditions
6. Use multiple conditions
7. Use `isin()`
8. Use `between()`
9. Filter text data
10. Combine filtering with Machine Learning datasets

These operations are very important for data analysis and
Machine Learning preprocessing.

```python
# Import Pandas
import pandas as pd
```

```python
# Create a sample student dataset

data = {
    "Name": ["Rahul", "Priya", "Amit", "Sneha", "Rohan", "Devi", "Jose"],
    "Age": [20, 21, 22, 20, 23, 34, 32],
    "Marks": [85, 90, 76, 92, 68, 100, 94],
    "Department": ["CSE", "ECE", "CSE", "CSE", "ECE", "CSE", "IT"]
}

df = pd.DataFrame(data)

print(df)
```

```python
# Select the Marks column
print(df["Marks"])
```

```python
# Select Name and Marks
print(df[["Name", "Marks"]])
```

## Selecting Rows with loc

```python
# Select the row with index label 2
print(df.loc[2])
```

```python
# Select rows 1 and 3
print(df.loc[[1, 3]])
```

```python
# Select the Name and Marks of row 1
print(df.loc[1, ["Name", "Marks"]]) # df.loc[rows, columns]
```

```python
# Select rows 1, 2 and 3
# and columns Name and Marks

print(df.loc[[1, 2, 3], ["Name", "Marks"]])
```

## Selecting Rows with iloc

```python
# Select the first row by position
print(df.iloc[0])
```

```python
# Select the third row by position
print(df.iloc[2])
```

```python
# Select the first column
print(df.iloc[:, 0]) # df.iloc[rows, columns]
```

```python
# Select the first two columns
print(df.iloc[:, 0:2])
```

# loc vs iloc

`loc` → label based

`iloc` → position based

Example:

df.loc[2]

means:

"Give me the row with index label 2."

df.iloc[2]

means:

"Give me the row at position 2."

With the default index they may look the same,
but they are conceptually different.

## Basic Filtering

```python
# Filter students whose marks are greater than 80
result = df[df["Marks"] > 80]

print(result)
```

```python
# Students with marks greater than or equal to 90
result = df[df["Marks"] >= 90]

print(result)
```

```python
# Students younger than 23
result = df[df["Age"] < 23]

print(result)
```

```python
# Students from CSE
result = df[df["Department"] == "CSE"]

print(result)
```

```python
# Students who are not from CSE
result = df[df["Department"] != "CSE"]

print(result)
```

## Multiple Conditions — AND

```python
# Filter using two conditions
result = df[(df["Age"] > 20) & (df["Marks"] > 80)]

print(result)
```

## Multiple Conditions — OR

```python
# Filter using OR
result = df[(df["Age"] > 30) | (df["Marks"] > 95)]

print(result)
```

## Combining AND and OR

```python
# CSE students with marks greater than 80
result = df[
    (df["Department"] == "CSE") &
    (df["Marks"] > 80)
]

print(result)
```

## isin()

```python
# Select students from CSE or IT
result = df[df["Department"].isin(["CSE", "IT"])]

print(result)
```

## NOT isin()

```python
# Select students who are not from CSE or IT
result = df[~df["Department"].isin(["CSE", "IT"])]

print(result)
```

## between()

```python
# Select students whose age is between 20 and 25
result = df[df["Age"].between(20, 25)]

print(result)
```

## Filtering Text

```python
# Find names containing the letter "a"
result = df[df["Name"].str.contains("a", case=False)]

print(result)
```

## Filtering Based on String Starting With

```python
# Select names beginning with "R"
result = df[df["Name"].str.startswith("R")]

print(result)
```

## Sorting After Filtering

```python
# Filter students and sort them by Marks
result = df[df["Marks"] > 80].sort_values("Marks")

print(result)
```

## Filter + Sort

```python
# Find students above 80 marks
# and sort them from highest to lowest

result = df[df["Marks"] > 80].sort_values(
    "Marks",
    ascending=False
)

print(result)
```

## Filtering and Selecting Specific Columns

```python
# Filter rows and select specific columns
result = df[
    (df["Department"] == "CSE") &
    (df["Marks"] > 80)
][["Name", "Marks"]]

print(result)
```

## Using loc for Filtering

```python
# Filter rows using loc
result = df.loc[
    (df["Department"] == "CSE") &
    (df["Marks"] > 80),
    ["Name", "Marks"]
]

print(result)
```
