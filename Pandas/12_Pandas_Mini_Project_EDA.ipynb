```python
# Import Pandas
import pandas as pd
```

```python
# Create an employee dataset

data = {
    "Employee_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    
    "Name": [
        "Rahul", "Priya", "Amit", "Sneha", "Rohan",
        "Devi", "Ankit", "Neha", "Arjun", "Puja"
    ],
    
    "Age": [
        22, 25, 28, 30, 35,
        40, 27, 29, 45, 32
    ],
    
    "Department": [
        "CSE", "ECE", "CSE", "CSE", "IT",
        "IT", "ECE", "CSE", "IT", "ECE"
    ],
    
    "Experience": [
        1, 3, 5, 7, 10,
        15, 4, 6, 20, 8
    ],
    
    "Salary": [
        25000, 35000, 45000, 50000, 70000,
        90000, 40000, 55000, 120000, 60000
    ],
    
    "Performance": [
        65, 72, 80, 85, 90,
        95, 75, 88, 98, 82
    ]
}

df = pd.DataFrame(data)

print(df)
```

## First Inspection

```python
# Display the first five rows
print(df.head())
```

```python
# Check the number of rows and columns
print(df.shape)
```

```python
# Check column names
print(df.columns)
```

```python
# Check data types and non-null values
df.info()
```

```python
# Get statistical information
print(df.describe())
```

```python
# Check missing values
print(df.isnull().sum())
```

```python
# Check for duplicate rows
print("Duplicate rows:", df.duplicated().sum())
```

```python
# Display unique departments
print(df["Department"].unique())
```

```python
# Count employees in each department
print(df["Department"].value_counts())
```

```python
# Calculate average salary
average_salary = df["Salary"].mean()

print("Average salary:", average_salary)
```

```python
# Find the highest salary
highest_salary = df["Salary"].max()

print("Highest salary:", highest_salary)
```

```python
# Find the lowest salary
lowest_salary = df["Salary"].min()

print("Lowest salary:", lowest_salary)
```

```python
# Find the employee with the highest salary

highest_salary_employee = df.loc[
    df["Salary"].idxmax()
]

print(highest_salary_employee)
```

```python
# Filter employees whose salary is greater than 60000

high_salary = df[df["Salary"] > 60000]

print(high_salary)
```

```python
# Employees with more than 5 years of experience

experienced = df[df["Experience"] > 5]

print(experienced)
```

```python
# Select employees from the CSE department

cse_employees = df[df["Department"] == "CSE"]

print(cse_employees)
```

```python
# CSE employees with more than 5 years of experience

result = df[
    (df["Department"] == "CSE") &
    (df["Experience"] > 5)
]

print(result)
```

```python
# Sort employees by salary from highest to lowest

result = df.sort_values(
    "Salary",
    ascending=False
)

print(result)
```

```python
# Calculate average salary by department

result = df.groupby("Department")["Salary"].mean()

print(result)
```

```python
# Generate a salary summary for each department

result = df.groupby("Department").agg(
    Average_Salary=("Salary", "mean"),
    Minimum_Salary=("Salary", "min"),
    Maximum_Salary=("Salary", "max"),
    Employee_Count=("Employee_ID", "count")
)

print(result)
```

```python
# Find employees with performance above 85

high_performers = df[
    df["Performance"] > 85
]

print(high_performers)
```

```python
# Display Name, Performance and Salary
# for high-performing employees

result = df.loc[
    df["Performance"] > 85,
    ["Name", "Performance", "Salary"]
]

print(result)
```

```python
# Create a new feature

df["Salary_Per_Experience"] = (
    df["Salary"] / df["Experience"]
)

print(df)
```

```python
# Display the new feature

print(
    df[
        ["Name", "Salary", "Experience", "Salary_Per_Experience"]
    ]
)
```

## Prepare X and y

```python
# Create the feature DataFrame

X = df[
    [
        "Age",
        "Department",
        "Experience",
        "Performance"
    ]
]

print(X)
```

```python
# Create the target

y = df["Salary"]

print(y)
```

## Check X and y

```python
# Check shapes

print("X shape:", X.shape)
print("y shape:", y.shape)
```

## Identify Numerical Features

```python
# Select numerical columns from X

numerical_features = X.select_dtypes(
    include=["int64", "float64"]
)

print(numerical_features)
```

## Identify Categorical Features

```python
# Select categorical columns from X

categorical_features = X.select_dtypes(
    include=["object"]
)

print(categorical_features)
```

```python
# Final dataset inspection

print("Shape:", df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nData types:")
print(df.dtypes)
```
