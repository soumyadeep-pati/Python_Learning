#  Pandas GroupBy and Aggregation

In this notebook, we will learn:

1. What `groupby()` does
2. Grouping data by one column
3. Calculating mean, sum, min and max
4. Counting records
5. Grouping by multiple columns
6. Using `agg()`
7. Combining `groupby()` with filtering
8. Understanding how GroupBy is useful in EDA and ML

```python
# Import Pandas
import pandas as pd
```

```python
# Create a student dataset

data = {
    "Name": [
        "Rahul", "Priya", "Amit", "Sneha",
        "Rohan", "Devi", "Jose", "Ankit"
    ],
    
    "Department": [
        "CSE", "ECE", "CSE", "CSE",
        "ECE", "CSE", "IT", "IT"
    ],
    
    "Gender": [
        "Male", "Female", "Male", "Female",
        "Male", "Female", "Male", "Male"
    ],
    
    "Age": [
        20, 21, 22, 20,
        23, 21, 22, 20
    ],
    
    "Marks": [
        85, 90, 76, 92,
        68, 100, 94, 80
    ]
}

df = pd.DataFrame(data)

print(df)
```

## Basic groupby()

```python
# Group the data by Department
grouped = df.groupby("Department")

print(grouped)
```

```python
# Calculate average marks for each department
average_marks = df.groupby("Department")["Marks"].mean()

print(average_marks)
```

## Sum

```python
# Calculate total marks for each department
total_marks = df.groupby("Department")["Marks"].sum()

print(total_marks)
```

## Minimum

```python
# Find the minimum marks in each department
minimum_marks = df.groupby("Department")["Marks"].min()

print(minimum_marks)
```

## Maximum

```python
# Find the maximum marks in each department
maximum_marks = df.groupby("Department")["Marks"].max()

print(maximum_marks)
```

## Count

```python
# Count the number of students in each department
student_count = df.groupby("Department")["Name"].count()

print(student_count)
```

## Multiple Statistics with agg()

```python
# Calculate multiple statistics for Marks
result = df.groupby("Department")["Marks"].agg(
    ["mean", "min", "max"]
)

print(result)
```

## Multiple Columns

```python
# Calculate the average Age and Marks for each department
result = df.groupby("Department")[["Age", "Marks"]].mean()

print(result)
```

## Multiple Aggregations on Multiple Columns

```python
# Calculate multiple statistics for multiple columns

result = df.groupby("Department").agg({
    "Age": ["mean", "min", "max"],
    "Marks": ["mean", "min", "max"]
})

print(result)
```

## GroupBy with count()

```python
# Count students in each department
result = df.groupby("Department").count()

print(result)
```

```python
result = df.groupby("Department")["Name"].count()

print(result)
```

## GroupBy with Multiple Categories

```python
# Group students by Department and Gender
result = df.groupby(
    ["Department", "Gender"]
)["Marks"].mean()

print(result)
```

## Reset the Index

```python
# Convert the grouped index back into a normal column

result = df.groupby("Department")["Marks"].mean().reset_index()

print(result)
```

## Sorting Grouped Results

```python
# Calculate average marks and sort from highest to lowest

result = (
    df.groupby("Department")["Marks"]
    .mean()
    .sort_values(ascending=False)
)

print(result)
```

## GroupBy + Filter

```python
average_marks = df.groupby("Department")["Marks"].mean()

print(average_marks)
```

```python
result = average_marks[average_marks > 80]

print(result)
```

## GroupBy + Sorting

```python
# Find average marks by department
# and sort from highest to lowest

result = (
    df.groupby("Department")["Marks"]
    .mean()
    .sort_values(ascending=False)
)

print(result)
```

## Named Aggregations

```python
# Create a summary with meaningful column names

result = df.groupby("Department").agg(
    Average_Marks=("Marks", "mean"),
    Highest_Marks=("Marks", "max"),
    Lowest_Marks=("Marks", "min"),
    Student_Count=("Name", "count")
)

print(result)
```
