#  Handling Missing Data with Pandas

In this notebook, we will learn:

1. What missing data means
2. How to detect missing values
3. How to count missing values
4. How to remove missing rows
5. How to remove missing columns
6. How to fill missing values
7. Filling with mean
8. Filling with median
9. Filling with mode
10. Forward fill and backward fill
11. Choosing the right strategy for ML

```python
# Import Pandas
import pandas as pd
```

```python
# Create a dataset containing missing values

data = {
    "Name": ["Rahul", "Priya", "Amit", "Sneha", "Rohan", "Devi"],
    "Age": [20, 21, None, 22, 23, None],
    "Marks": [85, None, 76, 92, None, 88],
    "Department": ["CSE", "ECE", "CSE", None, "ECE", "CSE"]
}

df = pd.DataFrame(data)

print(df)
```

## Detect Missing Values

```python
# Check which values are missing
print(df.isnull())
```

## Count Missing Values

```python
# Count missing values in each column
print(df.isnull().sum())
```

## Count All Missing Values

```python
# Count all missing values in the entire DataFrame
print(df.isnull().sum().sum())
```

```python
df.isnull()
```

```python
df.isna()
```

```python
print(df.isna().sum())
```

## Find Rows Containing Missing Values

```python
# Display rows containing at least one missing value
print(df[df.isnull().any(axis=1)])
```

## Find Rows With No Missing Values

```python
# Display rows that contain no missing values
print(df[df.notnull().all(axis=1)])
```

## Remove Rows Containing Missing Values

```python
# Create a copy of the original DataFrame
df_drop = df.copy()

# Remove rows containing missing values
df_drop = df_drop.dropna()

print(df_drop)
```

## Remove Rows Based on a Specific Column

```python
# Remove rows where Marks is missing
df_marks = df.dropna(subset=["Marks"])

print(df_marks)
```

## Remove Columns Containing Missing Values

```python
# Create a copy
df_columns = df.copy()

# Remove columns containing at least one missing value
df_columns = df_columns.dropna(axis=1)

print(df_columns)
```

## Filling Missing Values

```python
# Create a copy
df_fill = df.copy()

# Replace missing Age values with 0
df_fill["Age"] = df_fill["Age"].fillna(0)

print(df_fill)
```

## Fill With Mean

```python
# Create a copy
df_mean = df.copy()

# Calculate the mean Age
age_mean = df_mean["Age"].mean()

print("Mean Age:", age_mean)
```

```python
# Fill missing Age values with the mean Age
df_mean["Age"] = df_mean["Age"].fillna(age_mean)

print(df_mean)
```

```python
df_mean["Age"] = df_mean["Age"].fillna(df_mean["Age"].mean())
print(df_mean)
```

## Fill With Median

```python
# Create a copy
df_median = df.copy()

# Calculate the median Age
age_median = df_median["Age"].median()

print("Median Age:", age_median)
```

```python
# Fill missing Age values with the median
df_median["Age"] = df_median["Age"].fillna(age_median)

print(df_median)
```

```python
df_median["Age"] = df_median["Age"].fillna(df_median["Age"].median())
print(df_median)
```

## Fill Categorical Data with Mode

```python
# Find the most common department
department_mode = df["Department"].mode()[0]

print("Most common department:", department_mode)
```

```python
# Fill missing Department values with the mode
df_mode = df.copy()

df_mode["Department"] = df_mode["Department"].fillna(
    df_mode["Department"].mode()[0]
)

print(df_mode)
```

## Filling Different Columns Differently
    Age → numerical
    Marks → numerical
    Department → categorical

```python
# Create a copy
df_clean = df.copy()

# Fill numerical columns using median
df_clean["Age"] = df_clean["Age"].fillna(
    df_clean["Age"].median()
)

df_clean["Marks"] = df_clean["Marks"].fillna(
    df_clean["Marks"].median()
)

# Fill categorical column using mode
df_clean["Department"] = df_clean["Department"].fillna(
    df_clean["Department"].mode()[0]
)

print(df_clean)
```

## Forward Fill

```python
# Create a simple Series
values = pd.Series([10, None, None, 40, 50])

# Fill missing values using the previous value
print(values.ffill())
```

## Backward Fill

```python
# Fill missing values using the next available value
print(values.bfill())
```
