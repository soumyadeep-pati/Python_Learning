#  Pandas Series

A Pandas Series is a one-dimensional labeled data structure.

It can contain:

- Numbers
- Strings
- Boolean values
- Other Python objects

A Series contains:

1. Values
2. Index
3. Data type

Example:

    Index    Value
      0       10
      1       20
      2       30
      3       40

```python
# Import Pandas
import pandas as pd
```

```python
# Create a Series from a Python list
numbers = pd.Series([10, 20, 30, 40, 50])

print(numbers)
```

```python
# Display the index
print(numbers.index)
```

```python
# Display the values
print(numbers.values)
```

```python
numbers.index
```

```python
numbers.values
```

```python
# Access the first value
print(numbers[0])
```

```python
# Access the fourth value
print(numbers[3])
```

```python
# Create a Series with custom labels
marks = pd.Series(
    [85, 90, 78, 92],
    index=["Math", "Python", "DBMS", "ML"]
)

print(marks)
```

```python
# Access the marks for Python
print(marks["Python"])
```

```python
print(marks["ML"])
```

```python
# Create a dictionary
student_marks = {
    "Math": 85,
    "Python": 90,
    "DBMS": 78,
    "ML": 92
}

# Convert dictionary into a Series
marks = pd.Series(student_marks)

print(marks)
```

```python
# Select multiple subjects
print(marks[["Python", "ML"]]) # Notice the double brackets
```

```python
numbers = pd.Series([10, 20, 30, 40, 50])

# Get values from index 1 to index 3
print(numbers[1:4])
```

```python
numbers = pd.Series([10, 20, 30, 40])

print(numbers.dtype)
```

```python
prices = pd.Series([10.5, 20.5, 30.5])

print(prices.dtype)
```

```python
names = pd.Series(["Rahul", "Priya", "Amit"])

print(names.dtype)
```

```python
# Useful Series Methods
numbers = pd.Series([10, 20, 30, 40, 50])

# Number of values
print(numbers.size)
```

```python
# Get the largest value
print(numbers.max())
```

```python
# Get the smallest value
print(numbers.min())
```

```python
# Calculate the average
print(numbers.mean())
```

```python
# Calculate the sum
print(numbers.sum())
```

```python
# Calculate the standard deviation
print(numbers.std())
```

```python
numbers = pd.Series([10, 20, 30, 40, 50])

print(numbers.describe())
```

```python
marks = pd.Series([85, 90, None, 78, 92])

print(marks)
```

```python
# Check which values are missing
print(marks.isnull())
```

```python
# Count the number of missing values
print(marks.isnull().sum())
```

```python
# This is especially useful for categorical data.
results = pd.Series([
    "Pass",
    "Fail",
    "Pass",
    "Pass",
    "Fail",
    "Pass"
])

print(results.value_counts())
```

```python
# Basic Arithmetic with Series
marks = pd.Series([70, 80, 90, 60])

# Add 5 marks to every value
new_marks = marks + 5

print(new_marks)
```

```python
# Filtering a Series
marks = pd.Series([70, 80, 90, 60, 85])

# Select values greater than 75
print(marks[marks > 75])
```
