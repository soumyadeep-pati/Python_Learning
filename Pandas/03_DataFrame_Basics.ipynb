#  Pandas DataFrame Basics

A DataFrame is a two-dimensional data structure in Pandas.

It looks similar to:

- An Excel spreadsheet
- A database table
- A machine learning dataset

A DataFrame contains:

- Rows
- Columns
- Index
- Values

Example:

| Name | Age | Marks |
|------|-----|-------|
| Rahul | 20 | 85 |
| Priya | 21 | 90 |
| Amit | 20 | 78 |

In Machine Learning, datasets are commonly represented
using Pandas DataFrames.

```python
# Import the Pandas library
import pandas as pd
```

```python
# Create a dictionary containing student information
student_data = {
    "Name": ["Rahul", "Priya", "Amit", "Sneha", "Devi", "Ajay", "Jose", "Akash"],
    "Age": [20, 21, 20, 22, 34, 56, 32, 56],
    "Marks": [85, 90, 78, 92, 100, 76, 94, 67]
}

# Convert the dictionary into a DataFrame
df = pd.DataFrame(student_data)

# Display the DataFrame
print(df)
```

```python
# Get the number of rows and columns
print(df.shape)
```

```python
# Display all column names
print(df.columns)
```

```python
# Display the DataFrame index
print(df.index)
```

```python
# Display the data type of every column
print(df.dtypes)
```

```python
# Display the first five rows
print(df.head())
```

```python
# Display the first two rows
print(df.head(2))
```

```python
# Display the last five rows
print(df.tail())
```

```python
# Display the last two rows
print(df.tail(2))
```

```python
# Select the Age column
age = df["Age"]

print(age)
```

```python
# Select multiple columns
selected_data = df[["Name", "Marks"]]

print(selected_data)
```

```python
# Select the row with index label 0
print(df.loc[0])
```

```python
# Select the row with index label 2
print(df.loc[2])
```

```python
# Select rows with index labels 0 and 2
print(df.loc[[0, 2]])
```

```python
# Select Rahul's Marks
print(df.loc[0, "Marks"]) # df.loc[row, column]
```

```python
# Select Rahul's Name and Marks
print(df.loc[0, ["Name", "Marks"]])
```

```python
# Select rows 0 and 1 and columns Name and Marks
print(df.loc[[0, 1], ["Name", "Marks"]])
```

```python
# iloc means integer-location based indexing.
# It uses the numerical position of rows and columns.
# Select the first row
print(df.iloc[0])
```

```python
# Select the first column
print(df.iloc[:, 0]) 
```

```python
# Select the second column
print(df.iloc[:, 1])
```

```python
# Select the first three rows
print(df.iloc[0:3])
```

```python
# Select first three rows and first two columns
print(df.iloc[0:3, 0:2])
```

```python
# Filtering Data
# Select students with marks greater than 80
high_marks = df[df["Marks"] > 80]

print(high_marks)
```

```python
# Filter using multiple conditions
result = df[(df["Age"] > 20) & (df["Marks"] > 80)]

print(result)
```

```python
# Select students who are either older than 20
# OR have marks greater than 90
result = df[(df["Age"] > 20) | (df["Marks"] > 90)]

print(result)
```

```python
# Add a new column called Passed
df["Passed"] = df["Marks"] >= 40

print(df)
```

```python
# Create a new column containing marks after adding 5 bonus marks
df["Final_Marks"] = df["Marks"] + 5

print(df)
```

```python
# Remove the Final_Marks column
df = df.drop("Final_Marks", axis=1)

print(df) # Here: axis=1 means:column while: axis=0 means:row
```

```python
# Calculate basic statistics for the Marks column

print("Mean:", df["Marks"].mean())
print("Maximum:", df["Marks"].max())
print("Minimum:", df["Marks"].min())
print("Sum:", df["Marks"].sum())
```

```python
# Generate statistical information about numerical columns
print(df.describe())
```
