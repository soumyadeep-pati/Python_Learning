# 1. Tree Fundamentals

A **tree** is a hierarchical, non-linear data structure made of nodes connected by edges.

### Important terminology

- **Root**: topmost node
- **Parent**: a node directly above another node
- **Child**: a node directly below another node
- **Leaf**: node with no children
- **Edge**: connection between two nodes
- **Depth**: number of edges from the root to a node
- **Height**: number of edges on the longest path from a node to a leaf
- **Subtree**: a tree formed from a node and its descendants

Example:

```text
          10          ← Root
         /  \
        5    15       ← Children of 10
       / \ 
      2   7           ← Leaves
```


````python
class Node:
    def __init__(self, data):
        self.data = data
        self.children = []

root = Node(10)
root.children = [Node(5), Node(15)]

root.children[0].children = [Node(2), Node(7)]

print("Root:", root.data)
print("Children of root:", [node.data for node in root.children])
print("Children of 5:", [node.data for node in root.children[0].children])
````

# 2. Binary Tree

A **binary tree** is a tree where each node has at most two children:

- Left child
- Right child

```text
          10
         /  \
        5    15
       / \
      2   7
```

### Common types

1. **Full Binary Tree** — every node has either 0 or 2 children.
2. **Complete Binary Tree** — all levels are full except possibly the last, filled from left to right.
3. **Perfect Binary Tree** — all internal nodes have two children and all leaves are at the same level.
4. **Balanced Binary Tree** — left and right subtree heights stay reasonably close.
5. **Skewed Tree** — nodes mostly extend in one direction.


````python
class BinaryTreeNode:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

root = BinaryTreeNode(10)
root.left = BinaryTreeNode(5)
root.right = BinaryTreeNode(15)
root.left.left = BinaryTreeNode(2)
root.left.right = BinaryTreeNode(7)

print("Root:", root.data)
print("Left child:", root.left.data)
print("Right child:", root.right.data)
````

# 3. Tree Traversals

Traversal means visiting every node in a particular order.

The four important traversals are:

### DFS
- **Preorder:** Root → Left → Right
- **Inorder:** Left → Root → Right
- **Postorder:** Left → Right → Root

### BFS
- **Level Order:** visit nodes level by level

For this tree:

```text
          10
         /  \
        5    15
       / \
      2   7
```

Expected results:

- Preorder: `10 5 2 7 15`
- Inorder: `2 5 7 10 15`
- Postorder: `2 7 5 15 10`
- Level Order: `10 5 15 2 7`


````python
def preorder(root):
    if root is None:
        return
    print(root.data, end=" ")
    preorder(root.left)
    preorder(root.right)


def inorder(root):
    if root is None:
        return
    inorder(root.left)
    print(root.data, end=" ")
    inorder(root.right)


def postorder(root):
    if root is None:
        return
    postorder(root.left)
    postorder(root.right)
    print(root.data, end=" ")


root = BinaryTreeNode(10)
root.left = BinaryTreeNode(5)
root.right = BinaryTreeNode(15)
root.left.left = BinaryTreeNode(2)
root.left.right = BinaryTreeNode(7)

print("Preorder:")
preorder(root)

print("\nInorder:")
inorder(root)

print("\nPostorder:")
postorder(root)
````

## 3.1 Level Order Traversal — BFS

Level order uses a **queue**.

```text
          10
         /  \
        5    15
       / \
      2   7

10 → 5 → 15 → 2 → 7
```


````python
from collections import deque

def level_order(root):
    if root is None:
        return

    queue = deque([root])

    while queue:
        node = queue.popleft()
        print(node.data, end=" ")

        if node.left:
            queue.append(node.left)

        if node.right:
            queue.append(node.right)


level_order(root)
````

## 3.2 Iterative Traversals

Recursion uses the call stack. We can also implement DFS using an explicit stack.

### Iterative Preorder

Root → Left → Right

We push the right child first so the left child is processed first.


````python
def preorder_iterative(root):
    if root is None:
        return

    stack = [root]

    while stack:
        node = stack.pop()
        print(node.data, end=" ")

        if node.right:
            stack.append(node.right)

        if node.left:
            stack.append(node.left)


preorder_iterative(root)
````

````python
def inorder_iterative(root):
    stack = []
    current = root

    while stack or current:
        while current:
            stack.append(current)
            current = current.left

        current = stack.pop()
        print(current.data, end=" ")

        current = current.right


inorder_iterative(root)
````

# 4. Binary Search Tree (BST)

A **Binary Search Tree** is a binary tree satisfying:

```text
All values in left subtree < Root < All values in right subtree
```

Example:

```text
          50
         /  \
        30   70
       / \   / \
      20 40 60 80
```

This property makes searching efficient when the tree is balanced.

### Average-case complexity

| Operation | Average |
|---|---:|
| Search | O(log n) |
| Insert | O(log n) |
| Delete | O(log n) |

### Worst case

If the BST becomes skewed:

```text
10
  \
   20
     \
      30
        \
         40
```

then operations can become **O(n)**.


````python
class BST:
    def __init__(self):
        self.root = None

    def insert(self, root, key):
        if root is None:
            return BinaryTreeNode(key)

        if key < root.data:
            root.left = self.insert(root.left, key)
        elif key > root.data:
            root.right = self.insert(root.right, key)

        return root

    def search(self, root, key):
        if root is None or root.data == key:
            return root

        if key < root.data:
            return self.search(root.left, key)

        return self.search(root.right, key)


bst = BST()

for value in [50, 30, 70, 20, 40, 60, 80]:
    bst.root = bst.insert(bst.root, value)

print("Inorder traversal of BST:")
inorder(bst.root)

print("\nSearch 60:", bst.search(bst.root, 60).data)
````

## 4.1 BST Minimum and Maximum

In a BST:

- Minimum = keep moving **left**
- Maximum = keep moving **right**


````python
def find_min(root):
    current = root
    while current.left:
        current = current.left
    return current


def find_max(root):
    current = root
    while current.right:
        current = current.right
    return current


print("Minimum:", find_min(bst.root).data)
print("Maximum:", find_max(bst.root).data)
````

## 4.2 BST Deletion

There are three important cases.

### Case 1 — Leaf node

Simply remove it.

### Case 2 — One child

Replace the node with its child.

### Case 3 — Two children

Replace the node with its **inorder successor** (minimum value in the right subtree), then delete that successor.


````python
def delete_bst(root, key):
    if root is None:
        return None

    if key < root.data:
        root.left = delete_bst(root.left, key)

    elif key > root.data:
        root.right = delete_bst(root.right, key)

    else:
        # Case 1 and Case 2
        if root.left is None:
            return root.right

        if root.right is None:
            return root.left

        # Case 3: two children
        successor = find_min(root.right)
        root.data = successor.data
        root.right = delete_bst(root.right, successor.data)

    return root


print("Before deletion:")
inorder(bst.root)

bst.root = delete_bst(bst.root, 50)

print("\nAfter deleting 50:")
inorder(bst.root)
````

# 5. AVL Tree

An **AVL tree** is a self-balancing Binary Search Tree.

For every node:

```text
Balance Factor = Height(left subtree) - Height(right subtree)
```

A valid AVL tree must have:

```text
Balance Factor ∈ {-1, 0, +1}
```

If the balance factor becomes +2 or -2, a rotation is required.

### Four rotation cases

| Case | Fix |
|---|---|
| LL | Right Rotation |
| RR | Left Rotation |
| LR | Left Rotation + Right Rotation |
| RL | Right Rotation + Left Rotation |


````python
class AVLNode:
    def __init__(self, key):
        self.key = key
        self.left = None
        self.right = None
        self.height = 1


def height(node):
    if node is None:
        return 0
    return node.height


def update_height(node):
    node.height = 1 + max(height(node.left), height(node.right))


def balance_factor(node):
    if node is None:
        return 0
    return height(node.left) - height(node.right)
````

## 5.1 Right Rotation — LL Case

Before:

```text
        30
       /
      20
     /
    10
```

After right rotation:

```text
       20
      /  \
     10   30
```


````python
def right_rotate(y):
    x = y.left
    T2 = x.right

    x.right = y
    y.left = T2

    update_height(y)
    update_height(x)

    return x
````

## 5.2 Left Rotation — RR Case

Before:

```text
    10
      \
       20
         \
          30
```

After left rotation:

```text
       20
      /  \
     10   30
```


````python
def left_rotate(x):
    y = x.right
    T2 = y.left

    y.left = x
    x.right = T2

    update_height(x)
    update_height(y)

    return y
````

## 5.3 AVL Insertion

After normal BST insertion:

1. Update height.
2. Calculate balance factor.
3. Detect imbalance.
4. Apply the correct rotation.

The four cases are:

```text
LL → Right Rotation

RR → Left Rotation

LR → Left Rotation on left child,
     then Right Rotation

RL → Right Rotation on right child,
     then Left Rotation
```


````python
def avl_insert(root, key):
    # Normal BST insertion
    if root is None:
        return AVLNode(key)

    if key < root.key:
        root.left = avl_insert(root.left, key)
    elif key > root.key:
        root.right = avl_insert(root.right, key)
    else:
        return root

    # Update height
    update_height(root)

    # Check balance
    balance = balance_factor(root)

    # LL
    if balance > 1 and key < root.left.key:
        return right_rotate(root)

    # RR
    if balance < -1 and key > root.right.key:
        return left_rotate(root)

    # LR
    if balance > 1 and key > root.left.key:
        root.left = left_rotate(root.left)
        return right_rotate(root)

    # RL
    if balance < -1 and key < root.right.key:
        root.right = right_rotate(root.right)
        return left_rotate(root)

    return root
````

````python
def avl_inorder(root):
    if root:
        avl_inorder(root.left)
        print(root.key, end=" ")
        avl_inorder(root.right)


avl_root = None

for value in [30, 20, 10, 25, 28, 27]:
    avl_root = avl_insert(avl_root, value)

print("AVL inorder:")
avl_inorder(avl_root)

print("\nRoot:", avl_root.key)
print("Root balance factor:", balance_factor(avl_root))
````

# 6. Comparing Binary Tree, BST and AVL

| Feature | Binary Tree | BST | AVL |
|---|---|---|---|
| Max 2 children | Yes | Yes | Yes |
| Ordered | No | Yes | Yes |
| Self-balancing | No | No | Yes |
| Search worst case | O(n) | O(n) | O(log n) |
| Search average | O(n) | O(log n) | O(log n) |
| Insert worst case | O(n) | O(n) | O(log n) |

### Key idea

A BST can become skewed.

An AVL tree actively prevents excessive skew by performing rotations.
