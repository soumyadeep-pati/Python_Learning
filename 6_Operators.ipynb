# Python Operators

```python
# Arithmetic operators
x = 15
y = 4

print(x + y)
print(x - y)
print(x * y)
print(x / y)
print(x % y)
print(x ** y)
print(x // y)
```

```python
# Assignment operators
x = 10
x += 5
x -= 8
x *= 4
x /= 4
x //= 3
x %= 5
x **= 4
x &= 3
x |= 3 
```

```python
# Comparison operators
x = 5
y = 3

print(x == y)
print(x != y)
print(x > y)
print(x < y)
print(x >= y)
print(x <= y)
```

```python
# Ternary Operator
num = 6
x = "WEEKEND!" if num > 5 else "Workday"
print(x)

num = 6
x = "Fri" if num == 5 else "Sat" if num == 6 else "Sun" if num == 7 else "weekday"
print(x)
```

```python
# Logical operators
x = 5
print(x > 0 and x < 10)

x = 5
print(x < 5 or x > 10)

x = 5
print(not(x > 3 and x < 10))
```

```python
# Identity operators
x = ["apple", "banana"]
y = ["apple", "banana"]
z = x

print(x is z)
print(x is y)
print(x == y)
print(x is not y)
```

```python
# Membership operators
fruits = ["apple", "banana", "cherry"]

print("banana" in fruits)

print("pineapple" not in fruits)
```

```python
# Bitwise operators
print(6 & 3)
print(6 | 3)
print(6 ^ 3)
```
