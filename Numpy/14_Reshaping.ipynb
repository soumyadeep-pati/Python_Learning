# NumPy Reshaping

In this notebook, we will learn how to change the shape of a NumPy array.

Topics covered:
- reshape()
- flatten()
- ravel()
- Understanding shape

```python
import numpy as np
```

## 1. Creating an Array

Let's create a simple 1D array.

```python
arr = np.array([1, 2, 3, 4, 5, 6])

print(arr)
print("Shape:", arr.shape)
```

#  2. reshape()

`reshape()` changes the shape of an array without changing its data.

For example:

6 elements can be arranged as:
- 2 rows × 3 columns
- 3 rows × 2 columns

```python
arr = np.array([1, 2, 3, 4, 5, 6])

new_arr = arr.reshape(2, 3)

print(new_arr)
print("Shape:", new_arr.shape)
```

## 3. Another Example of reshape()

```python
arr = np.arange(1, 13)

new_arr = arr.reshape(3, 4)

print(new_arr)
print("Shape:", new_arr.shape)
```

## 4. flatten()

`flatten()` converts a multidimensional array into a 1D array.

It creates a new copy of the data.

```python
arr = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

flat = arr.flatten()

print("Original array:")
print(arr)

print("\nFlattened array:")
print(flat)
```

## 5. ravel()

`ravel()` also converts an array into 1D.

Unlike `flatten()`, `ravel()` usually returns a view when possible.

```python
arr = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

flat = arr.ravel()

print(flat)
```

## 6. reshape(-1)

`-1` tells NumPy to automatically calculate that dimension.

```python
arr = np.array([1, 2, 3, 4, 5, 6])

new_arr = arr.reshape(2, -1)

print(new_arr)
print("Shape:", new_arr.shape)
```

## Summary

- `reshape()` → changes array shape
- `flatten()` → converts to 1D and creates a copy
- `ravel()` → converts to 1D and usually returns a view
- `-1` → lets NumPy calculate the dimension automatically
