# NumPy Random Numbers

NumPy provides the `random` module for generating random numbers.

Random numbers are very important in Machine Learning,
especially for:
- Initializing model parameters
- Creating test data
- Simulations
- Sampling

```python
import numpy as np
```

## 1. Random Float

`np.random.rand()` generates random numbers between 0 and 1.

```python
number = np.random.rand()

print(number)
```

## 2. Multiple Random Numbers

```python
numbers = np.random.rand(5)

print(numbers)
```

## 3. Random Integers

`np.random.randint()` generates random integer values.

Syntax:

np.random.randint(start, end, size)

```python
numbers = np.random.randint(1, 10, size=5)

print(numbers)
```

## 4. Random Numbers in a Matrix

```python
matrix = np.random.randint(1, 100, size=(3, 3))

print(matrix)
```

## 5. Normal Distribution

`np.random.normal()` generates values following a normal distribution.

A normal distribution is commonly used in statistics and ML.

```python
numbers = np.random.normal(
    loc=0,
    scale=1,
    size=10
)

print(numbers)
```

## 6. Random Seed

Random numbers normally change every time we run the code.

A seed allows us to reproduce the same random numbers.

```python
np.random.seed(42)

print(np.random.rand(5))
```

## Summary

- `np.random.rand()` → random floats between 0 and 1
- `np.random.randint()` → random integers
- `np.random.normal()` → normally distributed values
- `np.random.seed()` → makes random results reproducible
