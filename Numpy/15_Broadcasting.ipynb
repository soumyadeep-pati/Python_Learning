# NumPy Broadcasting

Broadcasting allows NumPy to perform operations between arrays
of different but compatible shapes.

This is very important in Machine Learning because
we frequently perform mathematical operations on arrays.

```python
import numpy as np
```

## 1. Adding a Number to an Array

Normally, we might think we need a loop to add a number
to every element.

NumPy can do this automatically using broadcasting.

```python
arr = np.array([1, 2, 3, 4, 5])

result = arr + 10

print(result)
```

## 2. Multiplication

Broadcasting also works with multiplication.

```python
arr = np.array([1, 2, 3, 4, 5])

result = arr * 2

print(result)
```

## 3. Array + Array

Two arrays with the same shape can be added element by element.

```python
a = np.array([1, 2, 3])
b = np.array([10, 20, 30])

result = a + b

print(result)
```

## 4. Broadcasting with a 2D Array

A single number is broadcast to every element of the matrix.

```python
arr = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

result = arr + 10

print(result)
```

## 5. Broadcasting with a Row

```python
arr = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

row = np.array([10, 20, 30])

result = arr + row

print(result)
```

## Why Broadcasting Matters in ML

Broadcasting is commonly used for:

- Feature scaling
- Normalization
- Mathematical transformations
- Matrix calculations
- Neural network operations

Example:

If we have many data points and want to subtract the mean,
NumPy can perform the operation efficiently without writing loops.

## Summary

Broadcasting allows NumPy to perform operations on arrays
with compatible shapes without manually writing loops.
