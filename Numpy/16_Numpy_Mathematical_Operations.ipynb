# NumPy Mathematical Operations

In this notebook, we will learn common mathematical functions
used in data analysis and Machine Learning.

Topics:
- sum()
- mean()
- median()
- standard deviation
- min()
- max()

```python
import numpy as np
```

## 1. Creating Data

```python
data = np.array([10, 20, 30, 40, 50])

print(data)
```

## 2. Sum

`np.sum()` calculates the total of all elements.

```python
print(np.sum(data))
```

## 3. Mean

Mean is the average of the values.

Formula:

mean = sum of values / number of values

```python
print(np.mean(data))
```

## 4. Median

The median is the middle value after sorting the data.

```python
data = np.array([10, 30, 20, 50, 40])

print(np.median(data))
```

## 5. Minimum and Maximum

```python
print("Minimum:", np.min(data))
print("Maximum:", np.max(data))
```

## 6. Standard Deviation

Standard deviation tells us how spread out the values are
from the mean.

It is very important in statistics and Machine Learning.

```python
print("Standard deviation:", np.std(data))
```

## 7. Variance

Variance is related to standard deviation.

Standard deviation = square root of variance

```python
print("Variance:", np.var(data))
```

## 8. Mathematical Operations on 2D Arrays

```python
data = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print("Sum:", np.sum(data))
print("Mean:", np.mean(data))
print("Minimum:", np.min(data))
print("Maximum:", np.max(data))
```

## Summary

Important functions:

- `np.sum()` → total
- `np.mean()` → average
- `np.median()` → middle value
- `np.std()` → standard deviation
- `np.var()` → variance
- `np.min()` → minimum
- `np.max()` → maximum
