# 1. What is Machine Learning?

Machine Learning (ML) is a branch of computing in which algorithms
learn patterns from data and use those patterns to make predictions
or decisions.

Instead of explicitly writing every rule, we provide data from which
the algorithm can learn a useful relationship.

A simplified representation is:

    Data
      ↓
    Learning Algorithm
      ↓
    Model
      ↓
    Prediction

The central idea is:

    Learn from existing data
    and use the learned model
    on new data.
    

# 2. Traditional Programming vs Machine Learning

## Traditional Programming

In traditional programming, the programmer explicitly provides
the rules that transform inputs into outputs.

    Input + Rules → Output

Example:

    if marks >= 40:
        result = "Pass"
    else:
        result = "Fail"

The rules are written by the programmer.

---

## Machine Learning

In Machine Learning, we provide examples of inputs and outputs.

    Input + Output Examples
              ↓
       Learning Algorithm
              ↓
            Model

The model learns a relationship between the input and output.

After training:

    New Input
       ↓
    Trained Model
       ↓
    Prediction

```python
# Traditional programming example
# We explicitly define the rule for passing.

marks = 65

if marks >= 40:
    result = "Pass"
else:
    result = "Fail"

print("Marks:", marks)
print("Result:", result)
```

# 3. Why Do We Need Machine Learning?

Examples:

- Spam email detection
- Image recognition
- Speech recognition
- Recommendation systems
- Fraud detection
- Customer behavior prediction
- Medical prediction
- Demand forecasting

For some problems, writing every rule manually is difficult.

Machine Learning allows us to learn patterns from data.

The general idea is:

    Large / Complex Data
            ↓
      Learning Algorithm
            ↓
          Model
            ↓
       New Prediction

# 16. Machine Learning Model

A model is a mathematical function that maps input data
to an output.

Generally:

    y_hat = f(X)

where:

    X     = input
    f     = learned function/model
    y_hat = predicted output

For a simple linear model:

    y_hat = wx + b

where:

    w = weight
    b = bias

```python
# A very simple mathematical model

x = 5

# Model parameters
w = 10
b = 2

# Prediction
y_hat = w * x + b

print("Input:", x)
print("Weight:", w)
print("Bias:", b)
print("Prediction:", y_hat)
```

# 20. Actual Value and Predicted Value

Suppose:

    Actual value:
        y = 60

    Predicted value:
        y_hat = 52

The prediction is not perfect.

The error is:

    Error = y - y_hat

Therefore:

    Error = 60 - 52
          = 8

A Machine Learning algorithm needs a mathematical
way to measure this error.

```python
# Calculate prediction error

actual = 60
predicted = 52

error = actual - predicted

print("Actual value:", actual)
print("Predicted value:", predicted)
print("Error:", error)
```

# 21. Loss Function

A loss function measures how different the model's prediction
is from the actual target.

For a single regression example, one common loss is:

    L(y, y_hat) = (y - y_hat)^2

This is called the squared error.

Example:

    y = 60
    y_hat = 52

Then:

    L = (60 - 52)^2
      = 8^2
      = 64

The training process attempts to minimize the loss.

```python
# Calculate squared error

actual = 60
predicted = 52

squared_error = (actual - predicted) ** 2

print("Squared Error:", squared_error)
```

# 22. Mean Squared Error (MSE)

When there are multiple training examples,
we can calculate the average squared error.

The Mean Squared Error is:

                 1
    MSE =       --- Σ (y_i - y_hat_i)^2
                 n

where:

    n       = number of samples
    y_i     = actual value
    y_hat_i = predicted value

The objective is:

    Minimize MSE

This will become extremely important when we study
Linear Regression.

```python
# Example values

actual = np.array([10, 20, 30, 40])
predicted = np.array([12, 18, 29, 45])

# Calculate individual errors
errors = actual - predicted

# Square the errors
squared_errors = errors ** 2

# Calculate the average
mse = np.mean(squared_errors)

print("Actual values:", actual)
print("Predicted values:", predicted)
print("Errors:", errors)
print("Squared errors:", squared_errors)
print("MSE:", mse)
```

# 27. Overfitting

Overfitting occurs when a model learns the training data
too closely.

It may learn:

- Noise
- Random fluctuations
- Accidental patterns

Typical behavior:

    Training performance → Very good

    Test performance     → Poor

The model has failed to generalize properly.

# 28. Underfitting

Typical behavior:

    Training performance → Poor

    Test performance     → Poor

Therefore:

    Underfitting
        ↓
    Model too simple

    Good Generalization

# 29. Mathematical View of Learning

Suppose our model is:

    y_hat = f(X; θ)

where:

    X = input data
    θ = model parameters
    f = model
    y_hat = prediction

We define an objective/loss function:

    J(θ)

Training attempts to find parameters θ that minimize
the objective.

Mathematically:

    θ* = argmin J(θ)

where:

    θ* = optimal parameter values

This optimization idea appears throughout Machine Learning.

# 30. Optimization

Optimization means finding parameter values that minimize
or maximize an objective function.

In Machine Learning, we usually minimize a loss function.

For example:

    Minimize:

        J(θ)

The desired parameters are:

    θ* = argmin J(θ)

This concept is fundamental to many Machine Learning algorithms.

# 31. Gradient Descent

Gradient Descent is an optimization algorithm used to
minimize differentiable functions.

The general update rule is:

    θ_new = θ_old - α ∇J(θ)

where:

    θ      = model parameters
    α      = learning rate
    ∇J(θ)  = gradient of the objective function

The gradient indicates the direction in which the objective
increases most rapidly.

Therefore, we move in the opposite direction
to reduce the objective.

# 32. Learning Rate

The learning rate determines how large a step we take
during optimization.

It is represented by:

    α

If α is too large:

    The optimization may jump around
    or fail to converge.

If α is too small:

    Learning may become very slow.

Choosing an appropriate learning rate is therefore important
for gradient-based learning algorithms.

We will study this mathematically when we implement
Gradient Descent from scratch.

```python
# A simple one-dimensional gradient descent demonstration

# Objective function:
# J(theta) = theta^2
#
# Derivative:
# dJ/dtheta = 2 * theta

theta = 10.0
learning_rate = 0.1

for step in range(20):
    
    # Calculate gradient of J(theta) = theta^2
    gradient = 2 * theta
    
    # Update theta in the opposite direction of the gradient
    theta = theta - learning_rate * gradient
    
    print(f"Step {step + 1:2d}: theta = {theta:.6f}")
```

# 33. What Happened in Gradient Descent?

Our objective function was:

    J(θ) = θ²

Its derivative is:

    dJ/dθ = 2θ

The update rule is:

    θ_new = θ_old - α(2θ_old)

Starting from:

    θ = 10

the algorithm repeatedly moves θ toward:

    θ = 0

At:

    θ = 0

the function:

    J(θ) = θ²

reaches its minimum value:

    J(0) = 0

This simple example demonstrates the basic idea behind
gradient-based optimization.

```python
# Visualize the objective function and the optimization path

theta_values = np.linspace(-10, 10, 200)
loss_values = theta_values ** 2

# Store the path followed by gradient descent
theta = 10.0
learning_rate = 0.1

path = [theta]

for _ in range(20):
    gradient = 2 * theta
    theta = theta - learning_rate * gradient
    path.append(theta)

path = np.array(path)

plt.figure(figsize=(8, 5))

plt.plot(theta_values, loss_values, label="J(theta) = theta²")
plt.scatter(path, path**2, label="Gradient Descent Steps")

plt.xlabel("Theta")
plt.ylabel("Loss")
plt.title("Gradient Descent")

plt.legend()
plt.show()
```
