# Latent Factor Models

Latent Factor Models are statistical and machine learning models that explain observed data using a smaller number of hidden or **latent factors**.

The main idea is:

**Observed variables are generated from a smaller set of unobserved variables.**

For example, suppose we observe movie ratings from many users.

A user's rating for a movie may depend on hidden preferences such as:

- preference for action
- preference for comedy
- preference for romance
- preference for science fiction

These hidden preferences are called **latent factors**.

Instead of modeling every user-movie interaction independently, we try to discover a small number of hidden factors that explain the observed data.

A general latent factor model can be written as:

$$
X \approx ZW^T
$$

where:

- $X$ = observed data
- $Z$ = latent representation of observations
- $W$ = latent factor loadings
- $Z$ and $W$ contain a much smaller number of dimensions than $X$

The goal is to discover these hidden factors from observed data.

## 1. What Is a Latent Variable?

A **latent variable** is a variable that cannot be observed directly but influences the observed data.

For example, consider student performance.

We can directly observe:

- mathematics score
- physics score
- programming score
- statistics score

But we may believe that these scores are influenced by an underlying hidden ability such as:

**quantitative ability**

The quantitative ability is not directly observed, but it helps explain the observed scores.

Therefore:

$$
\text{Latent Factor}
\rightarrow
\text{Observed Variables}
$$

A latent factor model attempts to discover such hidden structure from the data.

### Examples of latent variables

| Application | Observed Data | Possible Latent Factors |
|---|---|---|
| Movies | User ratings | Genre preferences |
| Documents | Word frequencies | Topics |
| Finance | Asset returns | Market factors |
| Psychology | Survey responses | Personality traits |
| Images | Pixel values | Visual features |
| Recommendation systems | User-item ratings | User/item preferences |

The important idea is that the observed variables may have a much simpler underlying structure.

## 2. Why Latent Factor Models?

Real-world datasets often contain a large number of observed variables.

However, these variables are frequently correlated.

For example:

- mathematics score and physics score may be correlated
- action movies and superhero movies may receive similar ratings
- technology stocks may move together
- words such as "doctor", "hospital", and "medicine" may frequently occur together

This suggests that the data may be controlled by a smaller number of hidden factors.

Instead of representing an observation using $d$ variables, we can represent it using $k$ latent factors where:

$$
k \ll d
$$

This gives us a lower-dimensional representation of the data.

The main objectives are:

1. Discover hidden structure
2. Reduce dimensionality
3. Represent observations using latent factors
4. Reconstruct or predict observed data
5. Understand relationships between variables

## 3. Basic Mathematical Model

Suppose our observed data matrix is:

$$
X \in \mathbb{R}^{n \times d}
$$

where:

- $n$ = number of observations
- $d$ = number of observed variables

We assume that the data is generated using $k$ latent factors.

Let:

$$
Z \in \mathbb{R}^{n \times k}
$$

represent the latent representation of observations.

Let:

$$
W \in \mathbb{R}^{d \times k}
$$

represent the factor loadings.

The reconstructed data is:

$$
\hat{X} = ZW^T
$$

Therefore:

$$
X \approx ZW^T
$$

The important assumption is:

$$
k \ll d
$$

So instead of describing the data using $d$ dimensions, we describe it using only $k$ latent dimensions.

## 4. Understanding the Dimensions

Suppose:

$$
X \in \mathbb{R}^{1000 \times 50}
$$

This means:

- 1000 observations
- 50 observed variables

Suppose we choose:

$$
k = 5
$$

Then:

$$
Z \in \mathbb{R}^{1000 \times 5}
$$

and:

$$
W \in \mathbb{R}^{50 \times 5}
$$

The reconstruction becomes:

$$
\hat{X} = ZW^T
$$

with dimensions:

$$
(1000 \times 5)(5 \times 50)
=
1000 \times 50
$$

Therefore:

$$
\hat{X} \in \mathbb{R}^{1000 \times 50}
$$

The original data contains 50 dimensions, while the latent representation contains only 5 dimensions.

## 5. Generative View of Latent Factor Models

A useful way to understand latent factor models is through a generative process.

First, we assume that each observation has some hidden latent factors.

Then these latent factors generate the observed variables.

Conceptually:

$$
Z
\rightarrow
X
$$

More explicitly:

$$
\text{Latent Factors}
\rightarrow
\text{Observed Variables}
$$

For a single observation:

$$
x_i \approx Wz_i
$$

where:

- $z_i$ = latent representation
- $W$ = factor loading matrix
- $x_i$ = observed vector

The model therefore assumes that the observed data can be explained by a small number of hidden factors.

## 6. Example: Movie Preferences

Imagine that we have ratings from users for movies.

Suppose there are 6 movies:

- Action Movie
- Superhero Movie
- Romantic Movie
- Comedy Movie
- Drama Movie
- Horror Movie

Instead of treating every movie as completely independent, we may discover hidden factors such as:

- Action preference
- Romance preference
- Comedy preference

A user may have a latent representation such as:

$$
z_i =
\begin{bmatrix}
0.9 \\
0.1 \\
0.7
\end{bmatrix}
$$

This could represent:

- high action preference
- low romance preference
- high comedy preference

The model can then use these latent factors to reconstruct or predict movie ratings.

The factors do not necessarily have human-readable meanings.

They are mathematical dimensions that help explain the observed relationships.

## 7. Latent Factor Models vs PCA

Latent Factor Models and PCA are closely related, but they are not exactly the same.

### PCA

PCA searches for directions that explain maximum variance.

The representation is:

$$
X \approx ZW^T
$$

but PCA imposes specific constraints such as orthogonality of principal directions.

### Latent Factor Models

Latent factor models generally assume that observed variables are generated from hidden factors.

A common probabilistic formulation is:

$$
x = Wz + \mu + \epsilon
$$

where:

- $z$ = latent factor
- $W$ = loading matrix
- $\mu$ = mean vector
- $\epsilon$ = noise

The major difference is the probabilistic interpretation.

PCA is primarily a variance-maximization / linear dimensionality reduction technique.

Latent factor models provide a statistical model for how latent variables generate observations.

# 8. Factor Analysis

One of the most important latent factor models is **Factor Analysis**.

Factor Analysis assumes that observed variables are generated from a smaller number of latent factors plus noise.

The model is:

$$
x = \mu + \Lambda z + \epsilon
$$

where:

- $x$ = observed vector
- $\mu$ = mean vector
- $\Lambda$ = factor loading matrix
- $z$ = latent factor vector
- $\epsilon$ = noise

The latent factors are assumed to explain the correlations among the observed variables.

The noise term captures variation that is not explained by the latent factors.

## 9. Factor Analysis Equation

The model is:

$$
x = \mu + \Lambda z + \epsilon
$$

Assume:

$$
z \sim \mathcal{N}(0, I)
$$

and:

$$
\epsilon \sim \mathcal{N}(0, \Psi)
$$

where $\Psi$ is usually a diagonal covariance matrix.

Therefore:

$$
x \sim \mathcal{N}(\mu,\Lambda\Lambda^T+\Psi)
$$

This gives the covariance structure:

$$
\Sigma = \Lambda\Lambda^T+\Psi
$$

This equation is fundamental to Factor Analysis.

It says that the covariance of the observed variables is explained by two components:

$$
\text{Total Covariance}
=
\text{Common Factor Covariance}
+
\text{Unique Noise}
$$

or:

$$
\Sigma
=
\Lambda\Lambda^T
+
\Psi
$$

## 10. Common Variance and Unique Variance

Factor Analysis separates variation into two components.

### Common variance

Variation explained by shared latent factors.

For example, several exam scores may be influenced by a common underlying ability.

### Unique variance

Variation specific to an individual observed variable.

For example, a student's mathematics score may contain variation caused by factors that do not affect physics or programming scores.

Therefore:

$$
\text{Observed Variance}
=
\text{Common Variance}
+
\text{Unique Variance}
$$

This is one of the major conceptual differences between Factor Analysis and PCA.

PCA does not explicitly separate common variance from measurement-specific noise.

## 11. Factor Loading Matrix

The loading matrix $\Lambda$ describes how strongly each observed variable is related to each latent factor.

Suppose:

$$
\Lambda =
\begin{bmatrix}
0.9 & 0.1 \\
0.8 & 0.2 \\
0.2 & 0.9 \\
0.1 & 0.8
\end{bmatrix}
$$

There are:

- 4 observed variables
- 2 latent factors

The first two observed variables have high loadings on Factor 1.

The last two observed variables have high loadings on Factor 2.

Therefore, we may interpret:

$$
\text{Factor 1}
\rightarrow
\text{Variables 1 and 2}
$$

and:

$$
\text{Factor 2}
\rightarrow
\text{Variables 3 and 4}
$$

Factor loadings are useful for interpreting latent structure.

## 12. Factor Analysis with Python

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import FactorAnalysis
```

```python
iris = load_iris()

X = iris.data
feature_names = iris.feature_names

X.shape
```

## 13. Standardizing the Data

Factor analysis is generally sensitive to the scale of variables.

If variables have very different scales, variables with larger numerical ranges may dominate the model.

Therefore, standardization is often useful.

Standardization transforms each feature approximately to:

$$
x' = \frac{x-\mu}{\sigma}
$$

where:

- $\mu$ = feature mean
- $\sigma$ = feature standard deviation

After standardization:

$$
\text{mean} \approx 0
$$

and:

$$
\text{standard deviation} \approx 1
$$

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

X_scaled[:5]
```

## 14. Fit Factor Analysis

```python
factor_model = FactorAnalysis(
    n_components=2,
    random_state=42
)

X_factors = factor_model.fit_transform(X_scaled)

X_factors.shape
```

## 15. Latent Representation

The original dataset contains 4 observed variables.

After factor analysis, each observation is represented using only 2 latent factors.

Therefore:

$$
X \in \mathbb{R}^{150 \times 4}
$$

becomes:

$$
Z \in \mathbb{R}^{150 \times 2}
$$

The two columns of $Z$ represent the learned latent factors.

These factors summarize relationships among the original variables.

```python
latent_df = pd.DataFrame(
    X_factors,
    columns=["Factor 1", "Factor 2"]
)

latent_df.head()
```

## 19. Interpreting Factor Loadings

Factor loadings indicate the relationship between observed variables and latent factors.

A large absolute loading indicates a stronger relationship.

For example:

$$
|\lambda_{ij}| \approx 1
$$

suggests a strong relationship between variable $i$ and factor $j$.

A loading close to zero suggests a weak relationship.

However, latent factors do not automatically have meaningful names.

Interpretation requires examining which variables have large loadings on each factor.

### Important caution

Factor signs are not inherently meaningful.

A factor can often be multiplied by $-1$ without changing the underlying model.

Therefore:

$$
z \rightarrow -z
$$

and:

$$
\Lambda \rightarrow -\Lambda
$$

can represent essentially the same solution.

# 20. Choosing the Number of Latent Factors

Choosing the number of factors $k$ is an important modeling decision.

If $k$ is too small:

- important information may be lost
- reconstruction quality may be poor
- the model may underfit

If $k$ is too large:

- the model becomes unnecessarily complex
- factors may become difficult to interpret
- noise may be modeled as additional factors

Common approaches include:

1. Domain knowledge
2. Scree plots
3. Eigenvalue-based methods
4. Parallel analysis
5. Model selection criteria
6. Reconstruction or predictive performance
7. Interpretability

# 21. Latent Factor Models and Matrix Factorization

Latent factor models are closely connected to matrix factorization.

A matrix factorization represents:

$$
X \approx ZW^T
$$

where:

- $Z$ represents latent representations of observations
- $W$ represents latent representations of variables

This same idea appears in:

- recommender systems
- collaborative filtering
- topic modeling
- dimensionality reduction
- representation learning

The difference is mainly in the assumptions and learning procedure.

Matrix factorization often focuses on decomposing a matrix efficiently.

Probabilistic latent factor models additionally specify probability distributions and noise assumptions.

# 22. Probabilistic Matrix Factorization

A probabilistic latent factor model can be written as:

$$
x_{ij} = z_i^T w_j + \epsilon_{ij}
$$

where:

- $z_i$ = latent representation of observation $i$
- $w_j$ = latent representation of variable $j$
- $\epsilon_{ij}$ = noise

Assume:

$$
\epsilon_{ij} \sim \mathcal{N}(0,\sigma^2)
$$

Then:

$$
x_{ij} \sim
\mathcal{N}(z_i^T w_j,\sigma^2)
$$

The model therefore provides a probabilistic explanation for the observed values.

# 23. Latent Factor Models in Recommendation Systems

One of the most important applications of latent factor models is collaborative filtering.

Suppose we have a user-item rating matrix:

$$
R \in \mathbb{R}^{m \times n}
$$

where:

- $m$ = number of users
- $n$ = number of items

Many entries may be missing.

We assume:

$$
R \approx UV^T
$$

where:

$$
U \in \mathbb{R}^{m \times k}
$$

contains user latent factors.

And:

$$
V \in \mathbb{R}^{n \times k}
$$

contains item latent factors.

The predicted rating is:

$$
\hat{r}_{ij}
=
u_i^T v_j
$$

This allows the system to predict how much a user might like an item.

For example, the latent factors might capture hidden preferences such as:

- action vs romance
- popular vs niche
- serious vs humorous
- mainstream vs experimental

The factors are learned from the rating patterns rather than explicitly provided.

## 24. Learning Latent Factors

For observed entries, we can minimize reconstruction error.

A common objective is:

$$
\min_{U,V}
\sum_{(i,j)\in\Omega}
(r_{ij}-u_i^Tv_j)^2
$$

where:

- $\Omega$ = set of observed ratings
- $r_{ij}$ = actual rating
- $u_i^Tv_j$ = predicted rating

Regularization is often added:

$$
\min_{U,V}
\sum_{(i,j)\in\Omega}
(r_{ij}-u_i^Tv_j)^2
+
\lambda
\left(
\|U\|_F^2+\|V\|_F^2
\right)
$$

The regularization term helps prevent excessively large latent factors and reduces overfitting.

# 25. Building a Simple Latent Factor Model from Scratch

We can implement a simple matrix factorization model using gradient descent.

Suppose:

$$
R \approx UV^T
$$

For each observed rating:

$$
\hat{r}_{ij}=u_i^Tv_j
$$

The prediction error is:

$$
e_{ij}=r_{ij}-\hat{r}_{ij}
$$

We update the latent vectors using gradient descent.

```python
import numpy as np

R = np.array([
    [5, 4, 0, 1, 0],
    [4, 5, 0, 1, 2],
    [1, 0, 5, 4, 5],
    [0, 1, 4, 5, 4],
    [5, 4, 1, 0, 0]
], dtype=float)

R
```

## 26. Initialize Latent Factors

```python
np.random.seed(42)

num_users, num_items = R.shape
num_factors = 2

U = np.random.normal(
    scale=0.1,
    size=(num_users, num_factors)
)

V = np.random.normal(
    scale=0.1,
    size=(num_items, num_factors)
)

U.shape, V.shape
```

## 27. Training

```python
learning_rate = 0.01
regularization = 0.02
epochs = 1000

observed_positions = np.argwhere(R > 0)

loss_history = []

for epoch in range(epochs):

    total_loss = 0

    for i, j in observed_positions:

        prediction = np.dot(U[i], V[j])

        error = R[i, j] - prediction

        total_loss += error ** 2

        U_i = U[i].copy()
        V_j = V[j].copy()

        U[i] += learning_rate * (
            error * V_j - regularization * U_i
        )

        V[j] += learning_rate * (
            error * U_i - regularization * V_j
        )

    loss_history.append(total_loss)
```

## 28. Plot Training Loss

```python
plt.figure(figsize=(8, 5))

plt.plot(loss_history)

plt.xlabel("Epoch")
plt.ylabel("Reconstruction Error")
plt.title("Latent Factor Model Training Loss")

plt.show()
```

## 29. Reconstruct the Matrix

```python
R_pred = U @ V.T

R_pred
```

## 30. Compare Actual and Predicted

```python
print("Actual Rating Matrix:")
print(R)

print("\nPredicted Rating Matrix:")
print(np.round(R_pred, 2))
```

#   31. Predicting Missing Values

One of the major advantages of latent factor models is that they can estimate missing entries.

For example, if a user has not rated an item:

$$
r_{ij} = ?
$$

we can estimate it using:

$$
\hat{r}_{ij}=u_i^Tv_j
$$

This is the foundation of many collaborative filtering systems.

The model learns latent representations from known interactions and uses them to estimate unknown interactions.

```python
missing_positions = np.argwhere(R == 0)

predicted_missing = []

for i, j in missing_positions:
    predicted_missing.append(
        (i, j, R_pred[i, j])
    )

predicted_missing
```

# 32. Reconstruction Error

A basic way to evaluate a latent factor model is to measure reconstruction error.

For observed data:

$$
E =
\frac{1}{|\Omega|}
\sum_{(i,j)\in\Omega}
(r_{ij}-\hat{r}_{ij})^2
$$

This is the Mean Squared Error:

$$
MSE =
\frac{1}{|\Omega|}
\sum_{(i,j)\in\Omega}
(r_{ij}-\hat{r}_{ij})^2
$$

A smaller value indicates that the latent factors reconstruct the observed data more accurately.

However, low reconstruction error alone does not guarantee that the model generalizes well.

A validation or test set should be used when prediction is the objective.

```python
errors = []

for i, j in observed_positions:
    errors.append(
        (R[i, j] - R_pred[i, j]) ** 2
    )

mse = np.mean(errors)

print("Training MSE:", mse)
```

# 33. Regularization

Latent factor models can overfit, especially when the number of latent factors is large.

Regularization controls the magnitude of latent vectors.

A common objective is:

$$
L =
\sum_{(i,j)\in\Omega}
(r_{ij}-u_i^Tv_j)^2
+
\lambda
\left(
\|U\|_F^2+\|V\|_F^2
\right)
$$

where:

$$
\lambda > 0
$$

controls the strength of regularization.

### Small $\lambda$

The model has more freedom to fit the observed data.

### Large $\lambda$

The model is more strongly constrained.

The goal is to find a balance between:

**fit to data**

and

**model complexity**.

# 34. Effect of the Number of Latent Factors

The number of latent factors $k$ controls model capacity.

### Small $k$

Advantages:

- simpler model
- stronger compression
- easier visualization
- lower computational cost

Disadvantages:

- may lose important information
- may underfit

### Large $k$

Advantages:

- can capture more complex structure
- lower reconstruction error

Disadvantages:

- greater computational cost
- greater risk of overfitting
- harder interpretation

Therefore:

$$
k
$$

is an important hyperparameter.

# 35. Advantages of Latent Factor Models

### 1. Discover hidden structure

They can identify underlying factors that are not directly observed.

### 2. Dimensionality reduction

A high-dimensional dataset can be represented using fewer latent dimensions.

### 3. Handle sparse data

Matrix-based latent factor models are particularly useful when many observations are missing.

### 4. Useful for recommendation

They are fundamental to collaborative filtering.

### 5. Noise modeling

Probabilistic models can explicitly represent unexplained variation.

### 6. Flexible

Different likelihoods, priors, and factor structures can be used for different datasets.

### 7. Interpretability

Factor loadings can sometimes provide meaningful explanations of relationships between variables.

# 36. Disadvantages of Latent Factor Models

### 1. Choosing the number of factors

The correct value of $k$ is often not obvious.

### 2. Interpretability is not guaranteed

Latent factors are mathematical constructs and may not correspond to meaningful real-world concepts.

### 3. Rotational ambiguity

Different factor representations can explain the same observed covariance structure.

### 4. Can overfit

Too many latent factors can model noise.

### 5. Optimization can be difficult

Some models require iterative optimization and may have local minima.

### 6. Cold-start problem

Recommendation systems based on latent factors can struggle with completely new users or items.

### 7. Sensitive to assumptions

Probabilistic factor models depend on assumptions about distributions and noise.

# 27. Latent Factor Models vs Related Methods

| Method | Main Idea | Probabilistic? | Main Use |
|---|---|---:|---|
| PCA | Maximum variance directions | No | Dimensionality reduction |
| Factor Analysis | Hidden factors + noise | Yes | Discover latent structure |
| Matrix Factorization | Approximate matrix decomposition | Sometimes | Compression/recommendation |
| NMF | Non-negative factors | Usually no | Parts-based representation |
| Probabilistic Matrix Factorization | Probabilistic latent factors | Yes | Recommendation |
| Autoencoder | Neural network learns representation | Not necessarily | Nonlinear representation |

The methods are related because they all attempt to represent complex observations using lower-dimensional representations.

# 28. Latent Factor Models vs Autoencoders

Both latent factor models and autoencoders learn hidden representations.

### Latent Factor Model

A simple linear model is:

$$
X \approx ZW^T
$$

The relationship between latent variables and observations is linear.

### Autoencoder

An autoencoder learns:

$$
Z = f(X)
$$

and:

$$
\hat{X}=g(Z)
$$

where $f$ and $g$ are neural networks.

This allows autoencoders to learn nonlinear representations.

Therefore:

**Latent factor models → usually linear latent structure**

**Autoencoders → potentially nonlinear latent structure**

# 30. Important Assumptions

Different latent factor models make different assumptions.

For classical Factor Analysis, common assumptions include:

### 1. Latent factors explain correlations

Observed variables are correlated because they share latent causes.

### 2. Linear relationship

The basic model assumes:

$$
x = \mu + \Lambda z + \epsilon
$$

### 3. Latent factors have a specified distribution

Often:

$$
z \sim \mathcal{N}(0,I)
$$

### 4. Noise is separated from common factors

The residual term captures variable-specific variation.

### 5. Appropriate number of factors

The selected latent dimension should be sufficient to represent the underlying structure.

Violating these assumptions can reduce model quality.

# 31. Important Interpretation Warning

Latent factors should not automatically be interpreted as real-world causes.

Suppose a model discovers:

$$
Factor\ 1
$$

with high loadings on several variables.

It is tempting to give it a name such as:

**"Intelligence"**

However, the model itself only establishes a mathematical relationship.

A factor name is an interpretation made by the analyst.

Therefore:

**Correlation explained by a latent factor does not automatically imply causation.**

Domain knowledge is required to interpret latent factors responsibly.

x = \mu + \Lambda z + \epsilon
$$

This framework connects several important machine learning concepts:

**PCA → Factor Analysis → Matrix Factorization → Collaborative Filtering → Probabilistic Latent Models**

The key skill is not only learning how to fit the model, but also understanding:

- what the latent factors represent
- how many factors to use
- how to interpret factor loadings
- how to evaluate reconstruction
- how regularization controls complexity
- when latent factor models are appropriate
- when another representation-learning method is more suitable

Latent factor modeling is therefore an important bridge between classical statistical modeling, dimensionality reduction, recommender systems, and modern representation learning.
